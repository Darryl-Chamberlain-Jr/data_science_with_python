# Evaluation Overview

We have a few different methods for evaluating regression models. Scikit-Learn provides a list of scorers with pre-defined functions: https://scikit-learn.org/stable/modules/model_evaluation.html#scoring-string-names

You will see it is organized by model type: classification, regression, and clustering. We don't cover clustering in this course, but logistic regressions are used for classification and indeed sklearn.linear_model.LogisticRegression() uses accuracy as the default scoring method. 

One of the decisions a data scientist needs to make is how to evaluate their model. There is no one correct measure to use. The best you can do is know how the most common measures work and know how to implement other measures if the context requires it.

## Hypothesis Test on Model

The first step to evaluating a regression model is a hypothesis test on the entire model. The null hypothesis is that the null model  is the most predictive model with an alternative that at least one of the predictor variables included in the model creates a better fitting predictive model. In other words, the alternative hypothesis is:

&emsp; $H_0$: There is no linear/logistic relationship between the predictors $x_1, x_2, \ldots x_p \text{ and } y$. 

&emsp; $H_1$: There is some linear/logistic relationship between at least one of the predictors $x_1, x_2, \ldots x_p \text{ and } y$.


Like before, this can be translated to a statement about the coefficients for each predictor:

&emsp; $H_0: \, \beta_1 = \beta_2 = \ldots = \beta_p = 0$

&emsp; $H_1: \, \beta_k \neq 0 \text{ for some } k \in [1, p]$

Test Statistic for Regression Models
- Linear regression: F-statistic
- Logistic regression: Log-Likelihood Ratio

Technology will provide both the test statistic and the associated p-value. 

## Variance Explained by the Model (R-Squared)

For regression models, we can measure the proportion of variance explained by the model. In this case, variance is the amount of change that is likely if we tried to create the same model with a different data set. This measure will range between 0 and 1. A measure of 0 means we cannot explain any variance in the model, meaning the model will not be able to predict the response variable. A measure of 1 means we can explain all variance of the model, meaning the model can perfectly predict the response variable. For the purposes of our course, we would like the measure to be at least 0.7. There are three versions of this measure: $R^2$,  Adjusted $R^2$, and Pseudo $R^2$.

### Linear Regressions: R-Squared and Adjusted R-Squared

$R^2$ (also written as R-squared) is the proportion of variance explained by the linear regression model.

$$ R^2 = \frac{\text{TSS} - \text{RSS}}{\text{TSS}} $$

Where $\text{TSS } = \sum(y_i-\bar{y})^2$ is the Total Sum of Squares and $\text{RSS } = \sum (y_i - \beta_1 x_i - \beta_0)^2 $ is the Residual Sum of Squares.

Since TSS is the total variance in  and RSS is the amount of unexplained variance in , (TSS - RSS) is our explained variance. We then divide by TSS to scale it between 0 and 1.

Adjusted $R^2$ (also written as Adj. R-squared) adjusts for the number of predictor variables in the model by dividing each sum of squares by its degrees of freedom.

$$ \text{adjusted } R^2 = 
    \frac{
        \frac{ \text{TSS} }{ \text{df}_{\text{TSS}} } 
        - 
        \frac{ \text{RSS} }{ \text{df}_{\text{RSS}} }
    }
    {
        \frac{ \text{TSS} }{ \text{df}_{\text{TSS}} }
    } 
$$

But why adjust? Normally in a linear regression model, adding an additional predictor can only increase your . But ideally and realistically we want the least number of predictors to make a model. The adjusted R-Square works by penalizing the addition of predictors that do not substantially change the model's . We can then use adjusted R-Square to justify removing predictors that could be causing our model to overfit the data.  

### Logistic Regressions: Pseudo R-Squared

While a logistic regression does not have a "true" R-squared, there is a way to create a comparative measure called the Pseduo $R^2$. The calculation of this measure is beyond the scope of this course and you are expected to use technology to compute the Pseudo $R^2$. 

## Information Criterion

As we saw with the Adjusted R-Squared, some measures attempt to penalize the number of predictors as this can lead to models that overfit the training data and so do not generalize well for predictions on new sets of data. The information criterion measures attempt to balance the fit of a model on the training data with the model's complexity. 

$$\text{Information Criterion} = \text{fit} + \text{complexity}$$

$$ \text{fit } = -2 \ln(\text{likelihood}) $$

$$ \text{complexity } = \lambda k, \text{ where } k \text{ is the number of predictive variables }$$


The variable $\lambda$ is our complexity penalty. There are two frequent ways to penalize complexity.

In Akaike Information Criterion (AIC), $\lambda = 2$. AIC uses a flat penalty on the complexity of the model that does not take into account the number of data points you are using to create the regression model.

In Bayesian Information Criterion (BIC), $\lambda = \ln{N}$. BIC uses a varying penalty on the complexity of the model based on the number of data points $N$ you are using to create the regression model. 

Given two models fit on the same data set, the model with the smaller IC is better since it means the model is less complex while still being fit to the training data. Note that practically speaking, AIC and BIC frequently agree on which of two models is better. For our purposes you should report both and if there is disagreement, use BIC. 

## Common Classification Measures

Since logistic regressions model the probability of something occurring, they are frequently used to predict whether something will or will not happen. This classification opens up additional ways to measure logistic regressions based on how well they can predict the classification. In all of the cases below, a higher score is better. 

### Accuracy and Precision

Accuracy refers to how close an observed value is to a model value, while precision refers to the distribution of model values to some observed values. 

$$\text{accuracy} = \frac{\text{TP} + \text{TN}}{TP+FP+TN+FN} = \frac{\text{correct classifications}}{\text{all classifications}}$$

$$ \text{precision} = \frac{\text{TP}}{\text{TP}+\text{FP}} = \frac{\text{correct positive classifications}}{\text{all positive classifications}} $$

Accuracy is frequently used by software as the default measure for classification models. 

### Recall

Recall (also known as sensitivity) measures the proportion of true positives (positive model value and positive observed outcome) against all observed positive outcomes.

$$ \text{recall} = \frac{\text{TP}}{\text{TP}+\text{FN}} = \frac{\text{true positive classifications}}{\text{all positive outcomes}} $$

As you might expect, recall is important to consider when you want extreme accuracy in positive outcomes, such as in cases of positively identifying a disease or positively identifying a threat to an autonomous car. In these cases, the best model minimizes the frequency of the *worst* outcome: false negative (negative model value and positive observed outcome).

### F1 Score

If we want both Recall and Precision to be important to a model, we can use the F1 score, which is the harmonic mean of precision and recall. 

$$ \text{f1} = \frac{2 \cdot \text{precision} \cdot \text{recall}}{\text{precision} + \text{recall}} $$

F1 is useful in unbalanced data sets where there are more of one class for the response variable than another, such as in medical cases where we expect 80% of people tested to not have some disease.
